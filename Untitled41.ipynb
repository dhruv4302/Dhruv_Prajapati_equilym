```python
indicators = ['anc_coverage', 'institutional_delivery', 'immunization', 'high_risk_cases']
threshold_pct = 10.0

df_sorted = df.sort_values(['district', 'month'])

df_melted = df_sorted.melt(id_vars=['month', 'district'], value_vars=indicators, var_name='indicator', value_name='value')
df_melted['prev_value'] = df_melted.groupby(['district', 'indicator'])['value'].shift(1)
df_melted['pct_change'] = ((df_melted['value'] - df_melted['prev_value']) / df_melted['prev_value']) * 100

trends_df = df_melted[df_melted['pct_change'].abs() >= threshold_pct].dropna().copy()
trends_df['pct_change'] = trends_df['pct_change'].round(1)

print(trends_df[['district', 'indicator', 'month', 'prev_value', 'value', 'pct_change']].to_string(index=False))
```

```python
df_melted_outliers = df.melt(
    id_vars=['month', 'district'],
    value_vars=['anc_coverage', 'institutional_delivery', 'immunization', 'high_risk_cases'],
    var_name='indicator',
    value_name='value'
)

stats = df_melted_outliers.groupby(['month', 'indicator'])['value'].agg(
    q1=lambda x: x.quantile(0.25),
    q3=lambda x: x.quantile(0.75),
    state_mean='mean'
).reset_index()

stats['iqr'] = stats['q3'] - stats['q1']
stats['lower_bound'] = stats['q1'] - 1.5 * stats['iqr']
stats['upper_bound'] = stats['q3'] + 1.5 * stats['iqr']

outliers_merged = pd.merge(df_melted_outliers, stats, on=['month', 'indicator'])

outliers_df = outliers_merged[
    (outliers_merged['value'] < outliers_merged['lower_bound']) |
    (outliers_merged['value'] > outliers_merged['upper_bound'])
].copy()

outliers_df['state_mean'] = outliers_df['state_mean'].round(1)

print("Outliers:")
print(outliers_df[['district', 'indicator', 'month', 'value', 'state_mean', 'lower_bound', 'upper_bound']].to_string(index=False))
```

```python
indicators = ['anc_coverage', 'institutional_delivery', 'immunization', 'high_risk_cases']

corr_matrix = df[indicators].corr()

corr_unstacked = corr_matrix.unstack().reset_index()
corr_unstacked.columns = ['indicator_1', 'indicator_2', 'r_value']

flagged_corr = corr_unstacked[
    (corr_unstacked['indicator_1'] < corr_unstacked['indicator_2']) &
    (corr_unstacked['r_value'].abs() >= 0.70)
].copy()

flagged_corr['r_value'] = flagged_corr['r_value'].round(2)

corr_matrix.to_csv("correlation_matrix.csv")

print("=Correlational matrix:")
print(corr_matrix.round(2))

print("\n FLlagged pairs (|r| >= 0.70) ")
print(flagged_corr.to_string(index=False))
```

```python
df_sorted = df.sort_values(['district', 'month'])
df_melted_trend = df_sorted.melt(
    id_vars=['month', 'district'],
    value_vars=indicators,
    var_name='indicator',
    value_name='value'
)
df_melted_trend['prev_value'] = df_melted_trend.groupby(['district', 'indicator'])['value'].shift(1)
df_melted_trend['change_pct'] = ((df_melted_trend['value'] - df_melted_trend['prev_value']) / df_melted_trend['prev_value']) * 100

trends_ins = df_melted_trend[df_melted_trend['change_pct'].abs() >= 10.0].dropna().copy()
trends_ins['type'] = 'trend'
trends_ins['entity'] = trends_ins['district']
trends_ins['period'] = trends_ins['month']
trends_ins['change_pct'] = trends_ins['change_pct'].round(1)
trends_ins['severity'] = np.where(trends_ins['change_pct'].abs() >= 25.0, 'High', 'Medium')
trends_ins['explanation'] = (
    trends_ins['indicator'].str.replace('_', ' ').str.title() + " in " + trends_ins['entity'] +
    np.where(trends_ins['change_pct'] < 0, " dropped by ", " increased by ") +
    trends_ins['change_pct'].abs().astype(str) + "% compared to previous month."
)
```

```python
df_melted_out = df.melt(
    id_vars=['month', 'district'],
    value_vars=indicators,
    var_name='indicator',
    value_name='value'
)
stats = df_melted_out.groupby(['month', 'indicator'])['value'].agg(
    q1=lambda x: x.quantile(0.25),
    q3=lambda x: x.quantile(0.75),
    prev_value='mean'
).reset_index()
stats['iqr'] = stats['q3'] - stats['q1']
stats['lower_bound'] = stats['q1'] - 1.5 * stats['iqr']
stats['upper_bound'] = stats['q3'] + 1.5 * stats['iqr']

merged_out = pd.merge(df_melted_out, stats, on=['month', 'indicator'])
outliers_ins = merged_out[(merged_out['value'] < merged_out['lower_bound']) | (merged_out['value'] > merged_out['upper_bound'])].copy()
outliers_ins['type'] = 'outlier'
outliers_ins['entity'] = outliers_ins['district']
outliers_ins['period'] = outliers_ins['month']
outliers_ins['change_pct'] = (((outliers_ins['value'] - outliers_ins['prev_value']) / outliers_ins['prev_value']) * 100).round(1)
outliers_ins['prev_value'] = outliers_ins['prev_value'].round(1)
outliers_ins['severity'] = np.where(outliers_ins['change_pct'].abs() >= 25.0, 'High', 'Medium')
outliers_ins['explanation'] = (
    outliers_ins['entity'] + "'s " + outliers_ins['indicator'].str.replace('_', ' ') + " of " +
    outliers_ins['value'].astype(str) + " is an outlier compared to state mean (" +
    outliers_ins['prev_value'].astype(str) + ")."
)
```

```python
corrs_ins = pd.DataFrame({
    'type': ['correlation'] * len(flagged_corr),
    'indicator': flagged_corr['indicator_1'] + ":" + flagged_corr['indicator_2'],
    'entity': ['All Districts'] * len(flagged_corr),
    'period': [f"{df['month'].min()}..{df['month'].max()}"] * len(flagged_corr),
    'value': flagged_corr['r_value'],
    'prev_value': [np.nan] * len(flagged_corr),
    'change_pct': [np.nan] * len(flagged_corr),
    'severity': np.where(flagged_corr['r_value'].abs() >= 0.85, 'High', 'Medium'),
    'explanation': (
        "Strong correlation between " + flagged_corr['indicator_1'] + " and " + flagged_corr['indicator_2'] +
        " (r = " + flagged_corr['r_value'].astype(str) + "). Note: Fragile due to small sample size."
    )
})
```

```python
cols_order = ['type', 'indicator', 'entity', 'period', 'value', 'prev_value', 'change_pct', 'severity', 'explanation']
all_insights_df = pd.concat([trends_ins[cols_order], outliers_ins[cols_order], corrs_ins[cols_order]], ignore_index=True)
all_insights_df.insert(0, 'insight_id', [f"INS-{i+1:04d}" for i in range(len(all_insights_df))])
```

```python
all_insights_df.to_csv("generated_insights.csv", index=False)
```

```python
!pip install -q gradio

import gradio as gr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def render_dashboard(district_filter, month_filter, indicator_filter):
    df = pd.read_csv('district_health_data.csv')
    insights_df = pd.read_csv('generated_insights.csv')
    corr_matrix = pd.read_csv('correlation_matrix.csv', index_col=0)

    filtered_df = insights_df.copy()
    if district_filter != 'All':
        filtered_df = filtered_df[filtered_df['entity'].isin([district_filter, 'All Districts'])]
    if month_filter != 'All':
        filtered_df = filtered_df[filtered_df['period'].astype(str).str.contains(month_filter)]
    if indicator_filter != 'All':
        filtered_df = filtered_df[filtered_df['indicator'].astype(str).str.contains(indicator_filter)]

    display_table = filtered_df[['insight_id', 'type', 'entity', 'indicator', 'period', 'value', 'change_pct', 'severity', 'explanation']]

    fig1, ax1 = plt.subplots(figsize=(4.5, 3.2))
    if not filtered_df.empty:
        sev_counts = filtered_df['severity'].value_counts()
        sns.barplot(x=sev_counts.index, y=sev_counts.values, ax=ax1, palette='Set2')
        ax1.set_title("Insight Severity Distribution")
        ax1.set_ylabel("Count")
    else:
        ax1.text(0.5, 0.5, "No Data for Filters", ha='center', va='center')
    plt.tight_layout()

    fig2, ax2 = plt.subplots(figsize=(4.5, 3.2))
    ind_plot = indicator_filter if indicator_filter != 'All' else 'anc_coverage'
    sns.lineplot(data=df, x='month', y=ind_plot, hue='district', marker='o', ax=ax2)
    ax2.set_title(f"District Performance: {ind_plot}")
    ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize='x-small')
    plt.tight_layout()

    fig3, ax3 = plt.subplots(figsize=(4.5, 3.2))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", ax=ax3, cbar=False)
    ax3.set_title("Correlation Heatmap")
    plt.tight_layout()

    return display_table, fig1, fig2, fig3

raw_df = pd.read_csv('district_health_data.csv')
district_list = ['All'] + list(raw_df['district'].unique())
month_list = ['All'] + list(raw_df['month'].unique())
indicator_list = ['All', 'anc_coverage', 'institutional_delivery', 'immunization', 'high_risk_cases']

with gr.Blocks(title="Automated Insight Generation Engine") as demo:
    gr.Markdown("#  Automated Insight Generation Engine")
    gr.Markdown("District-level health indicators analysis with dynamic trends, outliers, and correlations.")

    with gr.Row():
        d_input = gr.Dropdown(choices=district_list, value='All', label='District Filter')
        m_input = gr.Dropdown(choices=month_list, value='All', label='Month Filter')
        i_input = gr.Dropdown(choices=indicator_list, value='All', label='Indicator Filter')

    insights_table = gr.Dataframe(label="Generated Insights (Section 4.1 Specification)")

    with gr.Row():
        chart_sev = gr.Plot(label="Severity Counts")
        chart_trend = gr.Plot(label="District Trend")
        chart_corr = gr.Plot(label="Correlation Matrix")

    ui_inputs = [d_input, m_input, i_input]
    ui_outputs = [insights_table, chart_sev, chart_trend, chart_corr]

    for input_widget in ui_inputs:
        input_widget.change(fn=render_dashboard, inputs=ui_inputs, outputs=ui_outputs)

    demo.load(fn=render_dashboard, inputs=ui_inputs, outputs=ui_outputs)

demo.launch(share=True, debug=False)
```
